```python
students = [
    {
        "name" : "Cols",
        "daily_exam" : 89,
        "weekly_exam" : 67,
        "grand_exam" : 56
    },
    {
        "name" : "Raena",
        "daily_exam" : 70,
        "weekly_exam" : 67,
        "grand_exam" : 67
    },
    {
        "name" : "kalisi",
        "daily_exam" : 90,
        "weekly_exam" : 45,
        "grand_exam" : 89
    },
    {
        "name" : "John",
        "daily_exam" : 78,
        "weekly_exam" : 56,
        "grand_exam" : 20
    },
    {
        "name" : "bartheon",
        "daily_exam" : 34,
        "weekly_exam" : 23,
        "grand_exam" : 30
    },
    {
        "name" : "Imp",
        "daily_exam" : 89,
        "weekly_exam" : 83,
        "grand_exam" : 76
    }
]

def avg_score(student):
    avg = (student["daily_exam"] + student["weekly_exam"] + student["grand_exam"]) / 3
    return avg

def classify(student):
    
    avg_s = avg_score(student)
    
    if  avg_s >=75:
        return "Ready"
    elif avg_s <  75 and avg_s >= 60:
        return "Improvement"
    else:
        return "High Risk"
for student in students:
    avg = avg_score(student)
    category = classify(student)
    print(f"{student["name"]}")
    print(f"Average : {avg: .2f}")
    print(f"Category : {category}")
    print("-----"*10)
    
# topper = max(students, key = avg-score)

topper=students[0]
for student in students:
    avg = avg_score(student)
    
    if avg >  avg_score(topper):
        topper=student
    
    
print("Topper")
print(f"Name : {topper["name"]}")
# print(f"Average : {round(avg_score(topper["name"]),2)}")
print(f"Average : {round(avg_score(topper),2)}")
print("-----"*10)

freq ={}
for student in students:
    category = classify(student)
    freq[category] = freq.get(category,0)+1
for key,val in freq.items():
    print(f"Category : {key} Count : {val}")
```
