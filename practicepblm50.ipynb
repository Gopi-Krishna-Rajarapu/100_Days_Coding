```python
class InsuficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self,account_number, holder_name,balance):
        self.__balance = balance
        self.holder_name = holder_name
        self.account_number = account_number
    def deposit(self,amount):
        
        if amount <= 0 :
            raise ValueError("Amount should  be positive")
        
        self.__balance += amount
        return f"Amount deposited :{amount} &  Successfully Deposited..!"
    def withdraw(self,amount):
        
        if amount > self.__balance:
            raise InsuficientBalanceError("Insufficient Balance")
        else:
            self.__balance -= amount
        return f"Amount withdrawn :{amount} &  Successfully Withdrawn..!"
    def display_balance(self):
         print("Account:", self.account_number)
         print("Holder:", self.holder_name)
         print("Balance:", self.__balance)

holder1 = BankAccount(123, "Robinson", 50000)
holder2 = BankAccount(124, "Richards", 650000)



transactions = [
    (holder1, "deposit", 23000),
    (holder2, "deposit", 150000),
    (holder1, "withdraw", 13000),
    (holder2, "withdraw", 22000),
    (holder1, "withdraw", 1000000),
    (holder1, "deposit", -500),
    (holder2, "withdraw", 0),

]
for holder,operation,amount in transactions:
    try:
        if operation == "withdraw":
            holder.withdraw(amount)
        elif operation == "deposit":
            holder.deposit(amount)
        
    except (InsuficientBalanceError,ValueError) as e:
        print(e)  
    # else:
    #     print(holder1.display_balance())
    #     print(holder2.display_balance())
    #     print(holder1.deposit(23000))
    #     print(holder2.deposit(150000))
    #     print(holder1.withdraw(13000))
    #     print(holder2.withdraw(22000))
    finally:
        print("Executed...!")
print(holder1.display_balance())
print(holder2.display_balance())      
```
