```python
class Solution:
    def isValid(self, s: str) -> bool:
        hash={')':'(',']':'[','}':'{'}
        stk=[]
        for i in s:
            if i not in hash:
                stk.append(i)
            else:
                if not stk:
                    return False
                else:
                    popped=stk.pop()
                    if popped!=hash[i]:
                        return False

        return not stk

          
    
obj= Solution()
print(obj.isValid('()[]{}'))
print(obj.isValid('(){[}]'))
```
