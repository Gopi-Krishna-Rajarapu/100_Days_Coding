```python

def sumOfDigits( n):
    total = 0

    for i in range(1, n + 1):
        num = i

        while num > 0:
            total += num % 10
            num //= 10

    return total
res=sumOfDigits(12)
print(res)
```

```python
# https://www.geeksforgeeks.org/problems/compute-sum-of-digits-in-all-numbers-from-1-to-n2855/1
class Solution:
    def sumOfDigits(self, n):
        total = 0
        factor = 1

        while factor <= n:
            higher = n // (factor * 10)
            current = (n // factor) % 10
            lower = n % factor

            total += higher * 45 * factor
            total += (current * (current - 1) // 2) * factor
            total += current * (lower + 1)

            factor *= 10

        return total
```
