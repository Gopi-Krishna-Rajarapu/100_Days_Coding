```python
def max_element(nums):
    freq = {}

    if len(nums)==1:
        return nums[0]
    # for i in nums:
    #     freq[i]=freq.get(i,0)+1
    for i in nums:
        freq[i] = freq.get(i, 0) + 1
    cnt=0
    maxele=float('-inf')
    for i,n in freq.items():
        if n>cnt:
            maxele=i
            cnt=n

    return maxele
max_element([2,2,1,1,1,2,2])   
```
