```python
class Solution:
    def reverse(self, x: int) -> int:
        rev=0
        rem=0
        num=abs(x)
        
       
        
        while num >0:
            rem=num%10
            rev=rev*10+rem
            num=num//10
        if x <0:
            rev=-rev
        if ((-2**31 > rev) or (rev> 2**31-1)):
            return 0
            
        return rev
obj1=Solution()
obj1.reverse(23931)
```
