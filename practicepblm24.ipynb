```python
def find_minimum(a, b):
    # code here  
   
    op_list=[a+b,a-b,a*b]
    try:
        op_list.append(a//b)
        
    except ZeroDivisionError:
        pass
    return min(op_list)

find_minimum(34,-3)
```
