```python

# You are given an integer, . Your task is to print an alphabet rangoli of size . 
# (Rangoli is a form of Indian folk art based on creation of patterns.)
n=5
alp=ord('a')
alp_1=alp-1
for i in range(n,0,-1):
    
    for j in range(2*i-2):
        print("-",end="")
    k=n
    while k >=i:
        prev=chr(alp_1 + k)
        print(chr(alp_1 + k),end="")
        if k!=i:
            print("-",end="")
        # print(prev.reverse())
        k-=1
    x=i+1    
    while x<=n:
        print("-",end="")
        print(chr(alp_1 + x),end="")
        x+=1
    for j in range(2*i-2):
        print("-",end="") 
    print()
    # 2-below part
    # print(\n)
    
for i in range(2,n+1):
    
    for j in range(i):
        print("-",end="")
    k=n
    while k >=i:
        prev=chr(alp_1 + k)
        print(chr(alp_1 + k),end="")
        if k!=i:
            print("-",end="")
        # print(prev.reverse())
        k-=1
    x=i+1    
    while x<=n:
        print("-",end="")
        print(chr(alp_1 + x),end="")
        x+=1
    for j in range(2*i-2):
        print("-",end="") 
    print()
```
