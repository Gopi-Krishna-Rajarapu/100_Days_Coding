```python
import string
sentence= input().lower().strip().translate(str.maketrans("","",string.punctuation))
d ={}

for ch in sentence.split():
    d[ch]= d.get(ch,0)+1
# for ch in inpt.split():
#     if ch in d:
#         d[ch] += 1
#     else:
#         d[ch] = 1
max_v=float('-inf')
max_w=''
for ch in d:  # Here d.items results tuple (key,value)
    if d[ch] > max_v:
        max_w = ch
        max_v = d[ch]
print(f"Highest frequency word: {max_w}")
print(f"Highest frequency word value: {max_v}")


sort_d= dict(sorted(d.items()))
print(f"Ascending or Sorted items: {sort_d}")

```
