```python
class Solution:
    def transpose(self, matrix: List[List[int]]) -> List[List[int]]:
        r = len(matrix)
        c = len(matrix[0])
        new_matrix = [[0] * r for _ in range(c)]
        for i in range(r):
            for j in range(c):
                new_matrix[j][i] = matrix[i][j]

        return new_matrix
obj=Solution()
obj.transpose([[1,2,3],[4,5,6],[7,8,9]])
```
