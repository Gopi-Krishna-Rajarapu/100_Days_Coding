```python
# Write a Python program to find and print the first prime number greater than 100. 
# Use a loop and the break statement to exit the loop once you find the prime number. 
def find_prime(first_num):
    s_num=first_num
    
    
    while (s_num<=200):
        if s_num < 1:
            s_num += 1
            continue
        if s_num >= 2  and (s_num & 1 == 0):
            s_num += 1
            continue
        i=2
        is_prime = True
        while i*i <= s_num:
            if s_num % i ==0 :
                is_prime = False
                break
            i += 1
        if is_prime :
            return s_num
        s_num +=1
    return None
res=find_prime(155)           
print(res)        

```

```python
# Modular 
def find_prime(first_num):
    if first_num <= 2:
        return 2

    n = first_num

    # Skip even numbers
    if (n & 1) == 0:
        n += 1

    while n <= 200:
        i = 3
        is_prime = True

        while i * i <= n:
            if n % i == 0:
                is_prime = False
                break
            i += 2

        if is_prime:
            return n

        n += 2

    return None


res = find_prime(123)
print(res)
```

```python
# Create a Python program to calculate the sum of all numbers from 1 to 100. 
# However, if the number is divisible by both 3 and 5, skip it using the continue statement. 
def cal_sum(f_num,l_num):
    s_num=0
    while (f_num <= l_num): 
        if (f_num % 3 == 0 ) or (f_num % 5 ==0) :
            f_num +=1
            continue
        else:
            s_num += f_num 
        f_num += 1
    return s_num
cal_sum(1,100)
```
