```python
print("Pattern-1",end="\t")
print("Right sliding")

n = 5
for i in range(n + 1):
    print(" " * (n - i) + "*" * i)


print("Patter-2",end="\t")
print("left sliding")

for i in range(n + 1):
    print("*" * (n - i))

# print("Pattern-3")
# print("Right sliding")
# for i in range(n+1):
#     for j in range(n-i):
#         print("*",end="")
#     # for k in range(i):
#     #     print("*",end="")
#     print()

print()
print("Pattern-3",end="\t")
print("Triangular pattern starting with 2")
n = 5

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i))
    

print()
print("Pattern-4",end="\t")
print("Triangular pattern starting with 1")
n = 5

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
print()

print("Pattern-5",end="\t")
print("Diamond")


for i in range(1, 2 * n):
    row = i if i <= n else 2 * n - i
    print(" " * (n - row) + "*" * (2 * row - 1))
    
    print()
```
